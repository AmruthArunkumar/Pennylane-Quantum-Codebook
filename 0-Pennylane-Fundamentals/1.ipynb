```python
import pennylane as qp
import numpy as np
```

```python
dev = qp.device("default.qubit", wires = 2)
```

```python
@qp.qnode(dev)
def circuit(phi): # Write any arguments you need here
    """
    This quantum function implements the circuit shown above
    and returns the output quantum state
    """

    qp.Hadamard(0)
    qp.PauliX(1)
    qp.CNOT([0, 1])
    qp.RY(phi, 1)

    return qp.state()
```

```python
def subcircuit_1(angle):
    qp.RX(angle, 0)
    qp.Y(1)

def subcircuit_2():
    qp.Hadamard(0)
    qp.CNOT([0, 1])

dev = qp.device('default.qubit', wires = 2)

@qp.qnode(dev)
def full_circuit(theta, phi):
    subcircuit_1(theta)
    subcircuit_2()
    subcircuit_1(phi)
    return qp.state()
```

```python
print(qp.draw(full_circuit)(0.1, 0.2))
```

```python
def subcircuit_1(angle, wire_list):
    qp.RX(angle, wire_list[0])
    qp.Y(wire_list[1])

def subcircuit_2(wire_list):
    qp.Hadamard(wire_list[0])
    qp.CNOT([wire_list[0], wire_list[1]])

dev = qp.device('default.qubit', wires = 2)

@qp.qnode(dev)
def full_circuit(theta, phi):
    subcircuit_1(theta, [0, 1])
    subcircuit_2([0, 1])
    subcircuit_1(phi, [1, 0])
    return qp.state()
```

```python
print(qp.draw(full_circuit)(0.1, 0.2))
```
